## Theory

### Introduction
The FIND-S algorithm is a basic concept learning algorithm in machine learning. It finds the most specific hypothesis that fits all the positive examples. The algorithm considers only positive training examples.

FIND-S starts with the most specific hypothesis and generalizes it each time it fails to classify an observed positive training data. Hence, it moves from the most specific hypothesis to the most general hypothesis.

### Important Representation
- `?` indicates that any value is acceptable for the attribute.
- A specific value (e.g., `Sunny`) indicates a single required value.
- `Φ` (phi) indicates that no value is acceptable.
- Most general hypothesis: `{?, ?, ?, ?, ?, ?}`
- Most specific hypothesis: `{Φ, Φ, Φ, Φ, Φ, Φ}`

### Algorithm
1. Initialize h to the most specific hypothesis in H.
2. For each positive training instance x:
   - For each attribute constraint aᵢ in h:
     - If the constraint aᵢ is satisfied by x, do nothing.
     - Else, replace aᵢ in h by the next more general constraint that is satisfied by x.
3. Output the hypothesis h.

### Worked Example
```
h0 = (Φ, Φ, Φ, Φ, Φ, Φ)
X1 = <Sunny, Warm, Normal, Strong, Warm, Same>  → Yes
h1 = <Sunny, Warm, Normal, Strong, Warm, Same>

X2 = <Sunny, Warm, High, Strong, Warm, Same>    → Yes
h2 = <Sunny, Warm, ?, Strong, Warm, Same>

X3 = <Rainy, Cold, High, Strong, Warm, Change>  → No (ignored)
h3 = <Sunny, Warm, ?, Strong, Warm, Same>

X4 = <Sunny, Warm, High, Strong, Cool, Change>  → Yes
h4 = <Sunny, Warm, ?, Strong, ?, ?>
```

## Procedure / Implementation

### Step 1 — Load the dataset

````python
import pandas as pd

df = pd.read_csv('play_tennis.csv')
print(df.head(10))
print('\nShape:', df.shape)
````

### Step 2 — Implement FIND-S

````python
attributes = df.columns[:-1]
target = df.columns[-1]
hypothesis = ['0'] * len(attributes)

for _, row in df.iterrows():
    if str(row[target]).strip().lower() == 'yes':
        if hypothesis == ['0'] * len(attributes):
            hypothesis = list(row[attributes])
        else:
            for i in range(len(attributes)):
                if hypothesis[i] != row[attributes[i]]:
                    hypothesis[i] = '?'

print('Final Hypothesis:', hypothesis)
````

### Step 3 — Show step-by-step trace

````python
hypothesis = ['0'] * len(attributes)
step = 0

for idx, row in df.iterrows():
    if str(row[target]).strip().lower() == 'yes':
        if hypothesis == ['0'] * len(attributes):
            hypothesis = list(row[attributes])
        else:
            for i in range(len(attributes)):
                if hypothesis[i] != row[attributes[i]]:
                    hypothesis[i] = '?'
        step += 1
        print(f'Step {step}: {hypothesis}')

print('\nFinal Maximally Specific Hypothesis:', hypothesis)
````

## Output / Observations
- FIND-S processes only positive examples and ignores negative ones.
- At each step the hypothesis is generalized by replacing mismatched attribute values with `?`.
- The final hypothesis is the most specific hypothesis consistent with all positive examples.
