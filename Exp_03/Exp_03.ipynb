## Procedure / Implementation

### Step 1 — Import libraries

```python
import pandas as pdimport matplotlib.pyplot as pltimport seaborn as sns
```

### Step 2 — Load the dataset

```python
df = pd.read_csv('titanic_train.csv')print(df.head())print('\nShape:', df.shape)print('\nInfo:')df.info()
```

### Step 3 — Check and clean the data

```python
print('Missing values:\n', df.isnull().sum())df['Age'] = df['Age'].fillna(df['Age'].median())df = df.drop_duplicates()print('\nAfter cleaning — missing Age:', df['Age'].isnull().sum())
```

### Step 4 — Transform the data

```python
df['Sex'] = df['Sex'].map({'male': 0, 'female': 1})df['FamilySize'] = df['SibSp'] + df['Parch'] + 1print(df[['Age', 'Sex', 'FamilySize']].head())
```

### Step 5 — Basic analysis and visualization

```python
print('Mean Age:', df['Age'].mean())print('Median Age:', df['Age'].median())sns.histplot(df['Age'], kde=True)plt.title('Age Distribution')plt.show()sns.countplot(x='Sex', data=df)plt.title('Gender Count')plt.show()
```

### Step 6 — Save the cleaned data

```python
df.to_csv('cleaned_titanic.csv', index=False)print('Cleaned dataset saved as cleaned_titanic.csv')
```

## Output / Observations- Displayed initial dataset and identified missing values (Age, Cabin, Embarked).- Filled missing Age with median, removed duplicate rows.- Encoded Sex as numeric and created a FamilySize feature.- Histograms and count plots show data distributions.- Saved a clean CSV ready for ML model training.
